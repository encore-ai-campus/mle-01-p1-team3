```python
import json
from langchain_core.documents import Document


with open("../data/RAG/maple_jobs.json", encoding="utf-8") as f:
    job_data = json.load(f)


def jobs_to_documents(jobs):
    documents = []

    for job in jobs:
        category = job.get("category", "")

        # 제논 - 직업군이 2개 이상인 경우 문자열로 변환
        if isinstance(category, list):
            category = ", ".join(category)

        document = Document(
            page_content=f"""
직업명: {job.get("name", "")}
별명: {job.get("subtitle", "")}
설명: {job.get("description", "")}
주스탯: {job.get("main_stat", "")}
무기: {job.get("weapon", "")}
""".strip(),
            metadata={
                "job_id": job.get("job_id", ""),
                "category": category,
                "name": job.get("name", ""),
                "url": job.get("url", ""),
            },
        )

        documents.append(document)

    return documents


job_documents = jobs_to_documents(job_data)

print(f"생성된 Document 수: {len(job_documents)}")
display(job_documents[:3])
```

```python
job_documents_dict = [
    {
        "page_content": document.page_content,
        "metadata": document.metadata,
    }
    for document in job_documents
]

with open(
    "../data/RAG/maple_jobs_documents.json",
    "w",
    encoding="utf-8",
) as f:
    json.dump(
        job_documents_dict,
        f,
        ensure_ascii=False,
        indent=2,
    )

print(
    f"저장된 Document 수: {len(job_documents_dict)}"
)
print(
    "저장 경로: ../data/RAG/maple_jobs_documents.json"
)
```
