```python
import json
import os
import re

BASE_PATH = "../data/processed"
```

```python
def clean_text(text):
    return re.sub(r"\s+", " ", text).strip()

def clean_nasted(data):
    # dict인 경우
    if isinstance(data, dict):
        return {
            key:clean_nasted(value)
            for key, value in data.items()
        }

    # list인 경우
    elif isinstance(data, list):
        return [
            clean_nasted(item) for item in data
        ]

    # 문자열인 경우
    elif isinstance(data, str):
        return clean_text(data)

    else:
        return data

def clean_url(items):
    seen_urls = set()
    result_reversed = []

    for item in reversed(items):
        url = item.get("url", "")

        if url:
            if url in seen_urls:
                continue

            seen_urls.add(url)

        result_reversed.append(item)

    return list(reversed(result_reversed))
```

```python
with open("../data/rag/maple_items.json", 'r', encoding='utf-8') as f:
    original_data = json.load(f)

print(len(original_data))

clean_data = clean_nasted(original_data)
clean_data = clean_url(clean_data)
display(clean_data[0])
```

```python
OUTPUT_PATH = os.path.join(BASE_PATH, "maple_items_process.json")

with open(OUTPUT_PATH, "w", encoding='utf-8') as f:
    json.dump(clean_data, f, ensure_ascii=False, indent=2)

print(OUTPUT_PATH)
```
