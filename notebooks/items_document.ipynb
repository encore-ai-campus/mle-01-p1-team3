```python
import json
from langchain_core.documents import Document
```

```python
with open("../data/processed/maple_items_process.json", 'r', encoding='utf-8') as f:
    item_data = json.load(f)

print(len(item_data))
display(item_data)
```

```python
def items_to_documents(items):
    documents = []

    for item in items:
        name = item.get("name", "")
        url = item.get("url", "")
        description = item.get("description", "")
        tables = item.get("tables", "")

        for table in tables:
            table_index = table.get("table_index")
            section_title = table.get("section_title")
            table_description = table.get("table_description")
            rows = table.get("rows", [])

            if not rows:
                continue

            # ['반지 레벨', '확률']
            header = rows[0]
            # ['1', '40%'] (인덱스 1부터)
            data_rows = rows[1:]

            for row_index, row in enumerate(data_rows, start=1):
                if not row:
                    continue

                if all(value is None or str(value).strip() == "" for value in row):
                    continue

                row_text = []
                
                for col_name, value in zip(header, row):

                    col_name = str(col_name).strip()
                    value = str(value).strip()

                    if value:
                        row_text.append(f"{col_name}: {value}")


                content_parts = []

                if name:
                    content_parts.append(f"문서: {name}")


                if section_title:
                    content_parts.append(f"섹션: {section_title}")

                if description:
                    content_parts.append(f"문서 설명: {description}")

                content_parts.extend(row_text)

                if table_description:
                    content_parts.append(f"참고사항: {table_description}")

                page_content = "\n".join(content_parts)

                doc = Document(
                    page_content=page_content,

                    metadata={
                        "source": "probability item",
                        "name": name,
                        "section_title": section_title,
                        "table_index": table_index,
                        "table_description": table_description,
                        "row_index": row_index,
                        "url": url
                    }
                )

                documents.append(doc)

                item_documents_dict = [
                    {
                        "page_content": doc.page_content,
                        "metadata": doc.metadata
                    }
                    for doc in item_documents
                ]

                with open("../data/RAG/maple_items_documents.json", "w", encoding='utf-8') as f:
                    json.dump(item_documents_dict, f, ensure_ascii=False, indent=2)

    return documents
```

```python
item_documents = items_to_documents(item_data)

print(f"생성된 Document 수: {len(item_documents)}")

display(item_documents)
```
