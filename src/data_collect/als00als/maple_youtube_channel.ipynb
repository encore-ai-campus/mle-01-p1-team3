```python
import os
from pathlib import Path

from dotenv import load_dotenv

env_path = Path(".env")
load_dotenv(env_path)

YOUTUBE_API_KEY = os.getenv("YOUTUBE_API_KEY")

if not YOUTUBE_API_KEY :
    raise ValueError(f"YOUTUBE_API_KEY was not found. Check this file: {env_path.resolve()}")
```

```python
import pandas as pd
import requests

chennel_url = "https://www.googleapis.com/youtube/v3/channels"

channel_params = {
    "part": "id",
    "forHandle" : "@메이플스토리맑음",
    "key": YOUTUBE_API_KEY,
}

channel_response = requests.get(chennel_url, params=channel_params)
channel_response.raise_for_status()
channel_video = channel_response.json()

display(channel_video)
```

```python
search_url = "https://www.googleapis.com/youtube/v3/search"

search_params = {
    "part" : "snippet",
    "channelId" : "UC1dHu9GhbHH7RcHKyJdaOvA",
    "order" : "date",
    "maxResults" : 20,
    "publishedAfter" : "2026-07-11T15:56:00Z",
    "publishedBefore" : "2026-08-11T15:56:00Z",
    "key" : YOUTUBE_API_KEY,
    "type" : "video"
}

search_response = requests.get(search_url, params=search_params)
search_response.raise_for_status()
search_video = search_response.json()

display(search_video)
```

```python
video_items = []
for video in search_video.get("items", []):
    video_list = video["id"]["videoId"]

    video_items.append({
        "videoId" : video_list
    })

# display(video_items)
from pathlib import Path
print(Path.cwd())

root_dir = Path.cwd().parent.parent

save_path = root_dir / "data" / "youtube_videos.csv"

video_df = pd.DataFrame(video_items)
display(video_df.head(5))
print(video_df.shape)

video_df.to_csv(save_path, encoding='utf-8')
```
