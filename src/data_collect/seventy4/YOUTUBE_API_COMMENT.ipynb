## YouTube Video Search


### 0. Load Environment Variables

```python
import os
from pathlib import Path

from dotenv import load_dotenv

env_path = Path(".env")
load_dotenv(env_path)

YOUTUBE_API_KEY = os.getenv("YOUTUBE_API_KEY")

if not YOUTUBE_API_KEY:
    raise ValueError(f"YOUTUBE_API_KEY was not found. Check this file: {env_path.resolve()}")
```

### 1. Search Videos with `search/list`


```python
import pandas as pd
import requests

search_url = "https://www.googleapis.com/youtube/v3/search"

query = "메이플육성", "메이플 가이드", "하이퍼버닝"
max_results = 100
```

```python
search_params = {
    "part": "snippet",
    "q": query,
    "type": "video",
    "maxResults": max_results,
    "key": YOUTUBE_API_KEY,
}

search_response = requests.get(search_url, params=search_params)
search_response.raise_for_status()
search_data = search_response.json()

search_items = []
for item in search_data.get("items", []):
    snippet = item.get("snippet", {})
    video_id = item.get("id", {}).get("videoId")
    if not video_id:
        continue

    search_items.append(
        {
            "video_id": video_id,
            "title": snippet.get("title"),
            "channel_title": snippet.get("channelTitle"),
            "published_at": snippet.get("publishedAt"),
            "description": snippet.get("description"),
        }
    )

search_df = pd.DataFrame(search_items)
display(search_df)
```

### 2. Get Video Details with `videos/list`


```python
videos_url = "https://www.googleapis.com/youtube/v3/videos"
```

```python
video_ids = search_df["video_id"].tolist()

videos_params = {
    "part": "snippet,statistics,contentDetails",
    "id": ",".join(video_ids),
    "key": YOUTUBE_API_KEY,
}

videos_response = requests.get(videos_url, params=videos_params)
videos_response.raise_for_status()
video_data = videos_response.json()

video_items = []
for item in video_data.get("items", []):
    snippet = item.get("snippet", {})
    statistics = item.get("statistics", {})
    content_details = item.get("contentDetails", {})

    video_items.append(
        {
            "video_id": item.get("id"),
            "video_title": snippet.get("title"),
            "category_id": snippet.get("categoryId"),
            "duration": content_details.get("duration"),
            "view_count": statistics.get("viewCount"),
            "like_count": statistics.get("likeCount"),
            "comment_count": statistics.get("commentCount"),
        }
    )

video_df = pd.DataFrame(video_items)
display(video_df)
```

### 3. Merge Search Results and Video Details


```python
youtube_df = search_df.merge(video_df, on="video_id", how="left")
display(youtube_df)
```

댓글과 대댓글을 추출하기 위해서는 유튜브의 CommentThreads API를 추가로 호출하는 과정이 필요합니다. 
기존 코드의 1~3단계를 그대로 유지한 상태에서, 
다음 단계(4단계, 5단계)로 댓글을 추출하고 CSV로 저장하는 코드를 덧붙여 수정 및 추가하시면 됩니다.

단계별 추가 및 수정 방법을 설명해 드리겠습니다.

1단계: 어떤 API가 필요한지 이해하기
댓글과 대댓글을 가져오기 위해서는 유튜브 공식 문서의 commentThreads 엔드포인트를 사용해야 합니다.
이 API는 영상의 최상위 댓글(Top-level comment)과 그 아래에 달린 대댓글(Replies)을 한 번에 가져올 수 있는 기능을 제공합니다.

2단계: 기존 코드 이어서 '댓글 수집' 셀 추가하기
기존 파일의 3. 영상 검색 정보와 상세 정보 데이터 병합 및 저장 코드 셀 아래에 새로운 코드 셀을 만들고 다음 코드를 입력하세요. 
1단계(검색 API)에서 구한 video_ids 리스트를 반복문으로 돌면서 댓글을 수집하는 로직입니다.

```python
# 4. 영상별 댓글 및 대댓글 추출
import requests
import pandas as pd
from IPython.display import display

# CommentThreads API 엔드포인트
comment_url = 'https://www.googleapis.com/youtube/v3/commentThreads'
comments_list = []

print("댓글 수집을 시작합니다...")

# 1단계에서 구한 영상 ID 리스트(video_ids)를 순회
for video_id in video_ids:
    params = {
        'part': 'snippet,replies', # snippet(최상위 댓글), replies(대댓글)
        'videoId': video_id,
        'maxResults': 100,         # 한 번에 가져올 최대 댓글 수 (최대 100)
        'key': YOUTUBE_API_KEY,
        'textFormat': 'plainText'  # HTML 태그 없는 일반 텍스트로 추출
    }
    
    while True:
        try:
            response = requests.get(comment_url, params=params, timeout=30)
            
            # 댓글이 비활성화되어 있거나 권한이 없는 영상은 건너뜀
            if response.status_code != 200:
                print(f"비디오 ID {video_id}의 댓글을 가져올 수 없습니다. (상태 코드: {response.status_code})")
                break
                
            data = response.json()
            
            for item in data.get('items', []):
                # 1) 최상위 댓글 추출
                top_comment = item['snippet']['topLevelComment']['snippet']
                comments_list.append({
                    'video_id': video_id,
                    'comment_id': item['id'],
                    'parent_id': None, # 최상위 댓글이므로 부모 ID 없음
                    'author': top_comment.get('authorDisplayName'),
                    'text': top_comment.get('textDisplay'),
                    'like_count': top_comment.get('likeCount'),
                    'published_at': top_comment.get('publishedAt'),
                    'is_reply': False # 대댓글 여부
                })
                
                # 2) 대댓글 추출 (있는 경우)
                if 'replies' in item:
                    for reply in item['replies']['comments']:
                        reply_snippet = reply['snippet']
                        comments_list.append({
                            'video_id': video_id,
                            'comment_id': reply['id'],
                            'parent_id': item['id'], # 부모 댓글(최상위 댓글)의 ID
                            'author': reply_snippet.get('authorDisplayName'),
                            'text': reply_snippet.get('textDisplay'),
                            'like_count': reply_snippet.get('likeCount'),
                            'published_at': reply_snippet.get('publishedAt'),
                            'is_reply': True # 대댓글 여부
                        })
            
            # 다음 페이지가 있으면 pageToken 업데이트, 없으면 반복 종료 (Pagination)
            if 'nextPageToken' in data:
                params['pageToken'] = data['nextPageToken']
            else:
                break
                
        except Exception as e:
            print(f"에러 발생: {e}")
            break

# 리스트를 데이터프레임으로 변환
comments_df = pd.DataFrame(comments_list)
print(f"총 {len(comments_df)}개의 댓글 및 대댓글을 수집했습니다.")
display(comments_df.head())
```

3단계: 추출한 데이터를 CSV 파일로 저장하기

마지막으로, 
위에서 만든 comments_df 데이터프레임을 csv 파일로 내보내는 코드를 작성합니다. 

한글 내용이 엑셀에서 깨지지 않게 하려면 encoding='utf-8-sig'를 반드시 넣어주어야 합니다.

```python
# 5. 수집한 댓글 데이터를 CSV 파일로 저장
import os

# 저장할 파일명 지정
file_name = "youtube_comments_result.csv"

# CSV로 저장 (한글 깨짐 방지를 위해 utf-8-sig 인코딩 사용)
comments_df.to_csv(file_name, index=False, encoding='utf-8-sig')

# 파일이 저장된 현재 절대 경로 출력
print(f"파일 저장 완료! 경로: {os.path.abspath(file_name)}")
```

API 호출 방식: 댓글은 [https://www.googleapis.com/youtube/v3/commentThreads](https://www.googleapis.com/youtube/v3/commentThreads) 주소로 호출해야 합니다.

part 파라미터: part='snippet,replies'로 설정해야 최상위 댓글(snippet)과 대댓글(replies)을 한 번에 받을 수 있습니다.
